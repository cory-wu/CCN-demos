# The Leaky Integrate-and-Fire Neuron

The Leaky Integrate-and-Fire (LIF) Neuron model is a fundamental mathematical representation of a biological neuron. It models a neuron's membrane potential as an electrical circuit that adds up incoming input while continuously leaking charge. If the neuron accumulates enough charge to hit some spiking threshold, it emits a spike and resets its membrane potential. The dynamics of the LIF neuron are given by
$$
C \dot\mu = \frac{\mu^0 - \mu(t)}{R} + I(t).
$$
The goal in this notebook is to use first principles to re-construct this equation.

An LIF neuron essentially does two things: it accumulates input over time, and it leaks its charge. To start, let's introduce some of the components of an LIF neuron.

Let $\mu(t)$ denote the membrane potential at time $t$. This represents charge accumulated across the cell membrane and measures how "close" a neuron is to firing.

Two forces act on $\mu(t)$ at every instant:

1. Leak: accumulated charge will "leak" out of the neuron since the membrane is not perfectly insulated. The rate of leak is determined by how far the current membrane potential is from an equilibrium resting potential. Let $\mu^0$ represent this membrane resting potential.
2. Inputs: incoming signals from other neurons or an external stimulus push the membrane potential toward the spiking threshold. The inputs feeding into the LIF neuron can be described with a time-indexed function, $I(t)$.

Armed with these three pieces, $\mu(t)$, $\mu^0$, and $I(t)$, we can define the fundamental core of an LIF neuron.

## LIF Fundamentals: Leak and Input

Using three pieces, we can define a basic differential equation that captures leak and input dynamics of the LIF neuron:
$$
\dot\mu = \mu^0 - \mu(t) + I(t),
$$
where $\dot\mu$ is the temporal derivative of the membrane potential $\mu(t)$. This equation posits that the change in membrane potential is dependent on a leak component and an input component, shown in red and green respectively.

$$\dot\mu = \underbrace{\color{red}{\mu^0 - \mu(t)}}_{\text{leak}} + \underbrace{\color{green}{I(t)}}_{\text{input}}$$

### Leak Demonstration

For now, let's consider the case where no input is being supplied to the neuron. That is, $I(t) = 0$ for all time steps $t$. In this case, the temporal derivative of $\mu(t)$ depends only on the leak term:
$$
\dot\mu = \mu^0 - \mu(t).
$$

This captures the biophysical leaky behavior of real neurons. The resting potential of a biological neuron is an electrochemical equilibrium between the ions inside of the cell and the ions outside of the cell. The neuron's membrane potential wants to be in equilibrium, so any deviation from the resting potential faces a restoring force that tries to bring the membrane potential back to rest.

The equation above reinforces $\mu^0$'s role as the membranes resting potential. If $\mu(t)$ is less than $\mu^0$, then $\dot\mu$ is positive and the membrane potential is increasing (toward $\mu^0$). If $\mu(t)$ is greater than $\mu^0$, then $\dot\mu$ is negative and the membrane potential is decreasing (still toward $\mu^0$). And if $\mu(t) = \mu^0$, then $\dot\mu = 0$ and the membrane potential is happy where it is; it does not change.

Let's start by setting the initial membrane potential above the resting potential, so $\mu(0) > \mu^0$. We'll see how the membrane potential converges to the resting potential.

So with no external input, any membrane potential above the resting potential eventually falls to the resting potential.

Likewise, any membrane potential below the resting potential will eventually rise to the resting potential.

<img src="animations/leak_below.gif" alt="Animated GIF" width="1000"/>

So the leak term in the differential equation of the LIF neuron acts to restore the neuron's membrane potential to its resting potential. This term is always active, representing the electrochemical forces that act on the neuron.

## Spiking

The simplest way to introduce a spiking mechanism for the LIF neuron is to add a conditional statement:
$$
\text{If my membrane voltage reaches $\theta$, then I emit a spike.}
$$
After spiking, the membrane potential *resets* to some value $\mu^{reset}$.

Let's see what the previous example looks like, but with the spiking mechanism added. Let spike threshold $\theta = -50$, and reset value $\mu^{reset} = -80$.

## Capacitance and Resistance

We now have a spiking neuron model that mimics biological neurons, and we are close to completing the full differential equation defined for the LIF neuron. There are two more components of the LIF neuron, the capacitance and resistance, that add additional flexibility to the temporal dynamics of the membrane potential. They manifest in the equation as:
$$
{\color{Green}{C}} \dot \mu = \frac{\mu^0 - \mu(t)}{\color{Red}{R}}  + I(t).
$$

*Capacitance* in a neuron represents how well it can store electrical charge. This determines how quickly a neuron's membrane potential responds to synaptic inputs. When the capacitance is high, then it takes more electrical charge to actually change the membrane's voltage. This results in a slower response to incoming signals.

This is captured in the equation above by scaling the temporal derivate $\dot\mu$. We can think of $C$ as a term that "eats up" the impact of input currents on $\dot\mu$. When $C$ is large, then the contribution of inputs $I(t)$ have less effect on $\dot\mu$ than if $C$ is small. We can see this by rearranging the equation to
$$
\dot\mu = \frac{\mu^0 - \mu(t)}{\color{Red}{R}\color{Green}{C}}\color{None} + \frac{I(t)}{\color{Green}{C}}
$$

If $C < 1$, then the membrane capacitance is small, and the membrane voltage responds even more quickly to synaptic inputs.

We'll start by a reasonable input current of 20 between $t = 2$ and $t=8$.

<img src="animations/input_step20.gif" alt="Animated GIF" width="1000"/>

Let's start by looking at small capacitance, say $C=0.5$. The membrane potential should react quickly to this input.

<img src="animations/spike_lowC.gif" alt="Animated GIF" width="1000"/>

With $C = 0.5$, we get 9 spikes with this input. Increasing the capacitance to something like $C = 2.0$, we'd expect to see fewer spikes, since much of input is able to be stored by the membrane. The membrane potential will react slower to the same input.

<img src="animations/spike_highC.gif" alt="Animated GIF" width="1000"/>

Now we only get 2 spikes with the same input!

In general, the membrane capacitance parameter $C$ allows us to modulate how sensitive the membrane potential is to external inputs. Low capacitance means more sensitivity, while high capacitance means less sensitivity. This additional flexibility brings us one step closer to viewing neurons as coincidence detectors (neurons that fire when many of its incoming inputs fire at the same time) or integrators (neurons that accumulate inputs over longer periods of time). The final piece of this puzzle is the membrane resistance, $R$.

When charges flow across membranes, they don't get to cross willy-nilly. They face *resistance*, which slows down their travel across the membrane. In neurons, the persistent leaking of ions across the cell membrane will also face resistance. With a high resistance, fewer ions will leak across the membrane. In the LIF neuron differential equation, the $R$ term captures this dynamic:
$$
C \dot \mu = \frac{\mu^0 - \mu(t)}{\color{Red}{R}} \color{None} + I(t).
$$

Recall that $\mu^0 - \mu(t)$ captures the leaky dynamics of the LIF neuron. Here, $R$ just scales the rate of this leak. High values of $R$ reduce the "leak" effect, while values of $R$ less than 1 scale up the "leak" effect.

Concretely, let's explore how different values of $R$ impact leaking dynamics. For this demonstration, we'll assume $I(t) = 0$, so there's no inputs to the neuron. Also, we'll assume $C = 1$.

<img src="animations/resistance_leak.gif" alt="Animated GIF" width="1000"/>

In this plot, it's clear how $R$ affects the leak dynamics of the neuron. Neurons with higher resistance take longer to "leak" back to equilibrium resting potential, compared to neurons with lower resistance.

Before we plot the results, we should expect that the lower-resistance neuron will leak charge faster. Alongside this constant input current, this neuron should spike more frequently than the higher-resistance neuron.

## A Note on the Membrane Time Constant

As mentioned before, in reality a neuron's input $I(t)$ is usually modeled as a combination of incoming inputs from neurons that synapse onto the neuron. One common way of modeling this is as a linear combination of spikes from presynaptic neurons (see eqn. $(2)$ in CCN).

The capacitance and resistance terms that are a part of the LIF equation allow us to capture two different "flavors" of neurons: coincidence detectors and integrators. Whether a particular neuron is a coincidence detector or an integrator depends on both its capacitance $C$ and its resistance $R$.

This manifests in the membrane time constant, commonly denoted as $\tau$ and defined as $\tau = RC$. To see where this comes from, note that if we multiply both sides by $R$,
$$
C \dot\mu = \frac{\mu^0-\mu(t)}{R} + I(t) \implies \tau \dot\mu = \mu^0 - \mu(t) + R\: I(t).
$$

Now, $\tau$ plays a similar role to capacitance above, by modulating how quickly the membrane potential relaxes towards equilibrium resting potential. This also allows us to distinguish coincidence detection and integration behaviors.


When $\tau$ is very small, then $\dot\mu$ becomes very sensitive to the leak and input terms. In the absence of any input, the membrane potential leaks very quickly toward resting potential. However, incoming input also quickly drives the membrane potential upward. This dynamic is poised for *coincidence detection*. A LIF neuron with small $\tau$ will never respond to a train of regular spikes spaced too far apart. The membrane potential will decay to resting potential between the spikes, so even though the neuron is receiving input, the structure of the input isn't suited for a response.

On the other hand, if $\tau$ is very large, then $\dot\mu$ is quite insensitive to leak and input terms. Without input, the membrane potential slowly leaks toward resting potential. Incoming input still pushes the membrane potential upward. This neuron is poised for *integration*, or the accumulation of inputs. A LIF neuron with a large $\tau$ will respond to a train of regular spikes *once enough spikes have been accumulated*. Though the membrane potential will decay between spikes, it decays slowly enough that each incoming spike contributes positively to the membrane potential, eventually pushing it over the spiking threshold.

We can see the fast fluctuations in the low time constant regime (left) and the integration of spikes in the high time constant regime (right).

## Conclusion

The Leaky Integrate-and-Fire Neuron has found its place as a popular abstraction of biological neurons. Its governing equation, which we've explored in this notebook, strikes a nice balance between biological plausibility and computational tractability.

To review the core components of LIF neuron:
- **Leaky Dynamics** - $\mu^0 - \mu(t)$ capture leaky behavior of neurons.
- **Flexible Inputs** - $I(t)$ is an arbitrary function that defines input current.
- **Time Constants** - Capacitance $C$ and Resistance $R$ form $\tau$, which modulates time-scales that the membrane potential acts on.
- **Spiking** - Spike Threshold $\theta$ defines a rule that introduces spiking behavior to the neuron.
